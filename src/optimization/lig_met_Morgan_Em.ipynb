```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import AllChem
from tqdm import tqdm
import pybel  # Open Babel Python bindings
import warnings
warnings.filterwarnings('ignore')

# Функция нормализации SMILES через Open Babel
def normalize_smiles_with_openbabel(smiles_str):
    """
    Нормализует SMILES строку с использованием Open Babel.
    Возвращает канонический SMILES с исправленной структурой или None при ошибке.
    """
    if pd.isna(smiles_str) or smiles_str is None or str(smiles_str).strip() == '':
        return None
    
    try:
        # Читаем SMILES через Open Babel
        mol = pybel.readstring("smi", str(smiles_str).strip())
        # Удаляем частичные заряды (если есть)
        mol.OBMol.DeletePartialCharges()
        # Устанавливаем ароматичность
        mol.OBMol.SetAromaticPerceived()
        # Конвертируем обратно в SMILES (канонический, с исправленной структурой)
        normalized_smiles = mol.write("smi", opt={"a": True, "h": True}).strip()
        # Удаляем точку с запятой и лишние пробелы
        normalized_smiles = normalized_smiles.split()[0].strip()
        return normalized_smiles
    except Exception as e:
        #print(f"Ошибка нормализации SMILES '{smiles_str}': {str(e)}")
        return None

# Функция для генерации Morgan fingerprint из SMILES
def smiles_to_morgan(smiles, radius=2, nBits=2048):
    """
    Конвертирует SMILES в Morgan fingerprint.
    Возвращает numpy array длиной nBits или нулевой вектор если SMILES некорректный/NaN
    """
    if pd.isna(smiles) or smiles is None or smiles.strip() == '':
        return np.zeros(nBits, dtype=int)
    
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol is None:
            return np.zeros(nBits, dtype=int)
        
        # Генерация Morgan fingerprint как битового вектора
        fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius, nBits=nBits)
        return np.array(fp)
    except Exception as e:
        #print(f"Ошибка при генерации fingerprint для '{smiles}': {str(e)}")
        return np.zeros(nBits, dtype=int)

# Основной код обработки
def process_smiles_and_generate_fps(df, output_normalized_file='normalized_smiles.csv'):
    """
    Основная функция обработки:
    1. Нормализует SMILES через Open Babel
    2. Сохраняет нормализованные SMILES в отдельный файл
    3. Генерирует Morgan fingerprints на основе нормализованных SMILES
    4. Добавляет fingerprints в исходный DataFrame
    
    Args:
        df: Исходный DataFrame
        output_normalized_file: Имя файла для сохранения нормализованных SMILES
    
    Returns:
        tuple: (DataFrame с fingerprints, DataFrame с нормализованными SMILES)
    """
    
    print("Начало обработки данных...")
    print(f"Исходный DataFrame: {df.shape[0]} строк, {df.shape[1]} колонок")
    
    # Создаем копию для нормализованных SMILES
    normalized_df = pd.DataFrame()
    normalized_df['index'] = df.index.copy()
    
    # Список столбцов для нормализации
    smiles_columns = {
        'lig_1': 'lig_1_normalized',
        'lig_2': 'lig_2_normalized', 
        'lig_3': 'lig_3_normalized',
        'Solvent_SMILES': 'solvent_normalized'
    }
    
    print("\n=== Этап 1: Нормализация SMILES через Open Babel ===")
    
    # Нормализуем каждый столбец
    for original_col, normalized_col in smiles_columns.items():
        if original_col in df.columns:
            print(f"Нормализация колонки '{original_col}'...")
            
            # Применяем нормализацию с прогресс-баром
            normalized_smiles_list = []
            for smiles in tqdm(df[original_col].values, desc=f"Нормализация {original_col}"):
                normalized_smiles = normalize_smiles_with_openbabel(smiles)
                normalized_smiles_list.append(normalized_smiles)
            
            # Сохраняем в оба DataFrame
            normalized_df[normalized_col] = normalized_smiles_list
            
            print(f"  - Успешно обработано: {len([x for x in normalized_smiles_list if x is not None])} из {len(normalized_smiles_list)}")
        else:
            print(f"Предупреждение: Колонка '{original_col}' не найдена в DataFrame")
    
    # Сохраняем нормализованные SMILES в отдельный файл
    print(f"\nСохранение нормализованных SMILES в файл: {output_normalized_file}")
    normalized_df.to_csv(output_normalized_file, index=False)
    print(f"Нормализованный dataset сохранен: {normalized_df.shape[0]} строк, {normalized_df.shape[1]} колонок")
    
    print("\n=== Этап 2: Генерация Morgan fingerprints на основе нормализованных SMILES ===")
    
    # Создаем копию исходного DataFrame для добавления fingerprints
    result_df = df.copy()
    
    # Параметры fingerprint
    nBits = 2048
    radius = 2
    print(f"Параметры Morgan fingerprint: radius={radius}, nBits={nBits}")
    
    # Обработка каждого типа молекул
    fingerprint_configs = [
        ('lig_1_normalized', 'lig_1', 'Morgan_1_'),
        ('lig_2_normalized', 'lig_2', 'Morgan_2_'),
        ('lig_3_normalized', 'lig_3', 'Morgan_3_'),
        ('solvent_normalized', 'Solvent_SMILES', 'Morgan_solv_abs_')
    ]
    
    for normalized_col, original_col, prefix in fingerprint_configs:
        if normalized_col in normalized_df.columns:
            print(f"\nГенерация fingerprints для '{original_col}' на основе нормализованных SMILES...")
            
            # Получаем нормализованные SMILES
            smiles_list = normalized_df[normalized_col].tolist()
            
            # Генерируем fingerprints
            fps = []
            for smiles in tqdm(smiles_list, desc=f"Генерация {prefix[:-1]}"):
                fp = smiles_to_morgan(smiles, radius, nBits) if smiles else np.zeros(nBits, dtype=int)
                fps.append(fp)
            
            # Добавляем колонки в result_df
            print(f"Добавление {nBits} колонок для '{prefix[:-1]}'...")
            for bit_idx in range(nBits):
                col_name = f"{prefix}{bit_idx}"
                result_df[col_name] = [int(fp[bit_idx]) for fp in fps]
            
            print(f"  - Создано колонок: {nBits}")
        else:
            print(f"Пропуск '{original_col}': нет нормализованных данных")
    
    print("\n=== Завершение обработки ===")
    print(f"Итоговый DataFrame: {result_df.shape[0]} строк, {result_df.shape[1]} колонок")
    
    # Статистика по созданным fingerprints
    created_fps = []
    for prefix in ['Morgan_1_', 'Morgan_2_', 'Morgan_3_', 'Morgan_solv_abs_']:
        cols = [col for col in result_df.columns if col.startswith(prefix)]
        if cols:
            created_fps.append(f"{prefix[:-1]}: {len(cols)} битов")
    
    if created_fps:
        print("\nСозданные fingerprint колонки:")
        for fp_info in created_fps:
            print(f"  - {fp_info}")
    
    return result_df, normalized_df

# Пример использования
if __name__ == "__main__":
    # Загрузка данных (замените на ваш реальный путь)
    # df = pd.read_csv('your_data.csv')
    
    # Пример данных для демонстрации
    print("Пример использования кода:")
    print("1. Загрузите ваш DataFrame:")
    print("   df = pd.read_csv('ваш_файл.csv')")
    print("\n2. Запустите обработку:")
    print("   result_df, normalized_df = process_smiles_and_generate_fps(df, 'normalized_smiles.csv')")
    print("\n3. Проверьте результаты:")
    print("   print(f'Исходные колонки: {df.columns.tolist()}')")
    print("   print(f'Колонки с fingerprints: {[col for col in result_df.columns if col.startswith(\"Morgan_\")][:5]}...')'")
    print("   print(f'Нормализованные SMILES сохранены в: normalized_smiles.csv')")
    
    # Дополнительные рекомендации
    print("\n=== Важные замечания ===")
    print("1. Убедитесь, что установлены необходимые пакеты:")
    print("   pip install pandas numpy rdkit openbabel tqdm")
    print("\n2. Для Open Babel может потребоваться установка системных зависимостей:")
    print("   sudo apt-get install openbabel (Linux)")
    print("   brew install open-babel (Mac)")
    print("   conda install -c conda-forge openbabel (Anaconda)")
    print("\n3. Для больших датасетов:")
    print("   - Рассмотрите возможность уменьшения nBits (например, до 1024)")
    print("   - Используйте параллельную обработку с joblib")
    print("   - Сохраняйте промежуточные результаты")
    print("\n4. Проверка качества нормализации:")
    print("   - Сравните оригинальные и нормализованные SMILES")
    print("   - Проверьте количество успешно обработанных молекул")
    print("   - Визуализируйте проблемные молекулы с помощью RDKit")
```

```python
import re
import pandas as pd
from openbabel import pybel

def normalize_smiles(smiles_str):
    """
    Нормализует SMILES: исправляет радикалы, заряды, изотопы,
    канонизирует структуру через OpenBabel.
    """
    if pd.isna(smiles_str):
        return None
    
    smiles_str = str(smiles_str).strip()
    
    # Пустая строка
    if not smiles_str or smiles_str == '':
        return None
    
    # === ИСПРАВЛЕНИЯ ПЕРЕД ОБРАБОТКОЙ ===
    
    # 1. Удаляем простые соли (они не несут структурной информации)
    # if smiles_str in {'[Cl-]', '[Br-]', '[I-]', '[F-]', '[O-]N=O'}:
    #     return None
    
    # 2. Исправляем явные радикалы и фрагменты
    radical_fixes = {
        '[H]': 'C',      # Водород → метан (минимальная органика)
        '[C]': 'C',      # Радикал → нейтральный углерод
        '[CH]': 'C',     # Метин-радикал → углерод
        '[CH2]': 'C',    # Метилен-радикал → углерод
        '[c-]':'C',
        '[C-]':'C',
        '[O-]':'O',
        '[O]':'O',
        '[CH3]': 'C',    # Метил-радикал → углерод
        '[Si]': '[Si]',  # Кремний оставляем (может быть валиден)
        '[C]1': 'C1',    # Радикал в цикле → нормальный углерод
        '[Cl-]':'[Cl]',
        '[C]#[O]':'C#O',
        '[Br--]':'[Br]',
        '[Br-2]':'[Br]',
        '[CH-]':'C',
        '[n-]':'N',
        '[N-]':'N',
        '[C-2]':'C',
        '[N]':'N',
        '[C@@]':'C@@',
        '[C@]':'C@',
    }
    
    for radical, fixed in radical_fixes.items():
        smiles_str = smiles_str.replace(radical, fixed)
    
    # 3. Исправляем странные формы CO (монооксид углерода)
    smiles_str = smiles_str.replace('[C-2]=O', 'C=O')  # Кетон
    smiles_str = smiles_str.replace('[C#O]', 'C#N')    # Нитрил (ближайший аналог)
    smiles_str = smiles_str.replace('[C]#[O]', 'C#N')
    
    # 4. Убираем заряды (преобразуем в нейтральные формы)
    charge_patterns = [
        (r'\[c-\]', 'c'),
        (r'\[n-\]', 'n'),
        (r'\[s-\]', 's'),
        (r'\[o-\]', 'o'),
        (r'\[O-\]', 'O'),
        (r'\[N-\]', 'N'),
        (r'\[C-\]', 'C'),
        (r'\[S-\]', 'S'),
        (r'\[C\+\]', 'C'),
        (r'\[N\+\]', 'N'),
        (r'\[O\+\]', 'O'),
        (r'\[n\+\]', 'n'),
        (r'\[CH2\]', 'C'),

    ]
    
    for pattern, replacement in charge_patterns:
        smiles_str = re.sub(pattern, replacement, smiles_str)
    
    # 5. Удаляем изотопные метки
    smiles_str = re.sub(r'\[(\d+)([A-Z][a-z]?)\]', r'[\2]', smiles_str)  # [2H] → [H]
    smiles_str = re.sub(r'\[2H\]', '[H]', smiles_str)
    smiles_str = re.sub(r'\[3H\]', '[H]', smiles_str)
    
    # 6. Удаляем только тяжёлые металлы (но оставляем органометаллы с углеродом)
    heavy_metals = ['Zr', 'Ta', 'Sn', 'Ag', 'As', 'Hg', 'Pb', 'Cd']
    metal_pattern = r'\[(' + '|'.join(heavy_metals) + r')[^\]]*\]'
    
    # Если SMILES - только металл, удаляем
    if re.match(r'^\[(' + '|'.join(heavy_metals) + r')[^\]]*\]$', smiles_str):
        return None
    
    # Если металл + органика, удаляем только металл
    smiles_str = re.sub(metal_pattern, '', smiles_str)
    
    # 7. Проверка после исправлений
    if not smiles_str or smiles_str.strip() == '':
        return None
    
    # === КАНОНИЗАЦИЯ ЧЕРЕЗ OPENBABEL ===
    try:
        mol = pybel.readstring("smi", smiles_str)
        mol.addh()  # Добавляем водороды
        
        # Канонический SMILES
        canonical_smiles = mol.write("can").strip().split()[0]
        
        return canonical_smiles
        
    except Exception as e:
        # Если OpenBabel не может обработать - возвращаем исправленный вариант
        print(f"Предупреждение: не удалось канонизировать '{smiles_str}': {e}")
        return smiles_str

# === ТЕСТЫ ===
test_cases = [
    '[Ir+2]([c-]1cc(C=O)ccc1-c1ccccn1)([c-]1cc(C=O)ccc1-c1ccccn1)(n1ccc(OC)cc1-c1cc(OC)ccn1)',  # Ваш пример с радикалом
]

for smi in test_cases:
    result = normalize_smiles(smi)
    print(f"{smi:40s} → {result}")
```

```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import AllChem
from tqdm import tqdm
import pybel  # Open Babel Python bindings
import re  # Добавлен импорт для регулярных выражений
import warnings
warnings.filterwarnings('ignore')
from copy import deepcopy

# === УЛУЧШЕННАЯ ФУНКЦИЯ НОРМАЛИЗАЦИИ SMILES ===
def normalize_smiles_with_openbabel(smiles_str):
    """
    Супер-функция нормализации SMILES:
    1. Исправляет радикалы, заряды, изотопы
    2. Удаляет тяжелые металлы и простые соли
    3. Канонизирует структуру через OpenBabel
    4. Возвращает валидный SMILES или None при ошибке
    """
    if pd.isna(smiles_str):
        return None
    smiles_str_ = deepcopy(smiles_str)
    smiles_str = str(smiles_str).strip()
    
    # Пустая строка
    if not smiles_str or smiles_str == '':
        return None
    
    # === ИСПРАВЛЕНИЯ ПЕРЕД ОБРАБОТКОЙ ===
    
    # 1. Удаляем простые соли (они не несут структурной информации)
    # if smiles_str in {'[Cl-]', '[Br-]', '[I-]', '[F-]', '[O-]N=O'}:
    #     return None
    
    radical_fixes = {
        '[H]': 'C',      # Водород → метан (минимальная органика)
        '[C]': 'C',      # Радикал → нейтральный углерод
        '[CH]': 'C',     # Метин-радикал → углерод
        '[CH2]': 'C',    # Метилен-радикал → углерод
        '[c-]':'C',
        '[C-]':'C',
        '[O-]':'O',
        '[O]':'O',
        '[CH3]': 'C',    # Метил-радикал → углерод
        '[Si]': '[Si]',  # Кремний оставляем (может быть валиден)
        '[C]1': 'C1',    # Радикал в цикле → нормальный углерод
        '[Cl-]':'[Cl]',
        '[C]#[O]':'C#O',
        '[Br--]':'[Br]',
        '[Br-2]':'[Br]',
        '[CH-]':'C',
        '[n-]':'N',
        '[N-]':'N',
        '[C-2]':'C',
        '[N]':'N',
        '[C@@]':'C@@',
        '[C@]':'C@',
    }
    
    for radical, fixed in radical_fixes.items():
        smiles_str = smiles_str.replace(radical, fixed)
    
    # 3. Исправляем странные формы CO (монооксид углерода)
    smiles_str = smiles_str.replace('[C-2]=O', 'C=O')  # Кетон
    smiles_str = smiles_str.replace('[C#O]', 'C#N')    # Нитрил (ближайший аналог)
    smiles_str = smiles_str.replace('[C]#[O]', 'C#N')
    
    # 4. Убираем заряды (преобразуем в нейтральные формы)
    charge_patterns = [
        (r'\[c-\]', 'c'),
        (r'\[n-\]', 'n'),
        (r'\[s-\]', 's'),
        (r'\[o-\]', 'o'),
        (r'\[O-\]', 'O'),
        (r'\[N-\]', 'N'),
        (r'\[C-\]', 'C'),
        (r'\[S-\]', 'S'),
        (r'\[C\+\]', 'C'),
        (r'\[N\+\]', 'N'),
        (r'\[O\+\]', 'O'),
        (r'\[n\+\]', 'n'),
        (r'\[CH2\]', 'C'),

    ]
    
    for pattern, replacement in charge_patterns:
        smiles_str = re.sub(pattern, replacement, smiles_str)
    
    # 5. Удаляем изотопные метки
    smiles_str = re.sub(r'\[(\d+)([A-Z][a-z]?)\]', r'[\2]', smiles_str)  # [2H] → [H]
    smiles_str = re.sub(r'\[2H\]', '[H]', smiles_str)
    smiles_str = re.sub(r'\[3H\]', '[H]', smiles_str)
    
    # 6. Удаляем только тяжёлые металлы (но оставляем органометаллы с углеродом)
    heavy_metals = ['Zr', 'Ta', 'Sn', 'Ag', 'As', 'Hg', 'Pb', 'Cd','Na']
    metal_pattern = r'\[(' + '|'.join(heavy_metals) + r')[^\]]*\]'
    
    # Если SMILES - только металл, удаляем
    if re.match(r'^\[(' + '|'.join(heavy_metals) + r')[^\]]*\]$', smiles_str):
        return None
    
    # Если металл + органика, удаляем только металл
    smiles_str = re.sub(metal_pattern, '', smiles_str)
    
    # 7. Проверка после исправлений
    if not smiles_str or smiles_str.strip() == '':
        return smiles_str_
    
    # === ЭТАП 2: КАНОНИЗАЦИЯ ЧЕРЕЗ OPENBABEL ===
    try:
        # Парсим SMILES через OpenBabel
        mol = pybel.readstring("smi", smiles_str)
        
        # Критические проверки валидности
        if mol.OBMol.NumAtoms() == 0:
            return None
        
        # Добавляем водороды и стандартизуем структуру
        mol.addh()
        mol.OBMol.SetAromaticPerceived()  # Явно устанавливаем ароматичность
        mol.OBMol.DeletePartialCharges()   # Удаляем частичные заряды
        
        # Генерируем канонический SMILES
        canonical_smiles = mol.write("can", opt={"h": True}).strip()
        canonical_smiles = canonical_smiles.split()[0]  # Удаляем комментарии
        
        # Финальная валидация RDKit
        if Chem.MolFromSmiles(canonical_smiles) is None:
            return smiles_str_
            
        return canonical_smiles
        
    except Exception:
        # Безопасный fallback: пробуем предобработанный SMILES
        if Chem.MolFromSmiles(smiles_str) is not None:
            return smiles_str
        return smiles_str_

# Функция для генерации Morgan fingerprint из SMILES
def smiles_to_morgan(smiles, radius=2, nBits=2048):
    """
    Конвертирует SMILES в Morgan fingerprint.
    Возвращает numpy array длиной nBits или нулевой вектор если SMILES некорректный/NaN
    """
    if not smiles or pd.isna(smiles):
        return np.zeros(nBits, dtype=np.int8)
    
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol is None:
            return np.zeros(nBits, dtype=np.int8)
        
        # Оптимизированная генерация fingerprint
        fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius, nBits=nBits)
        return np.array(fp, dtype=np.int8)
    except Exception:
        return np.zeros(nBits, dtype=np.int8)

# Основной код обработки
def process_smiles_and_generate_fps(df, output_normalized_file='normalized_smiles.csv'):
    """
    Основная функция обработки:
    1. Нормализует SMILES через улучшенную супер-функцию
    2. Сохраняет нормализованные SMILES в отдельный файл
    3. Генерирует Morgan fingerprints на основе нормализованных SMILES
    4. Добавляет fingerprints в исходный DataFrame
    
    Args:
        df: Исходный DataFrame
        output_normalized_file: Имя файла для сохранения нормализованных SMILES
    
    Returns:
        tuple: (DataFrame с fingerprints, DataFrame с нормализованными SMILES)
    """
    
    print("🚀 Начало обработки данных...")
    print(f"📊 Исходный DataFrame: {df.shape[0]} строк, {df.shape[1]} колонок")
    
    # Создаем копию для нормализованных SMILES
    normalized_df = pd.DataFrame()
    normalized_df['index'] = df.index.copy()
    
    # Список столбцов для нормализации
    smiles_columns = {
        'L1': 'lig_1_normalized',
        'L2': 'lig_2_normalized', 
        'L3': 'lig_3_normalized',
        'Solvent_SMILES': 'solvent_normalized'
    }
    
    print("\n✨ === Этап 1: Продвинутая нормализация SMILES ===")
    
    # Нормализуем каждый столбец
    stats = {}
    for original_col, normalized_col in smiles_columns.items():
        if original_col not in df.columns:
            print(f"⚠️  Предупреждение: Колонка '{original_col}' не найдена в DataFrame")
            continue
            
        print(f"\n🔧 Нормализация колонки '{original_col}'...")
        valid_count = 0
        normalized_list = []
        
        # Обработка с прогресс-баром
        for smiles in tqdm(df[original_col].values, desc=f"Обработка {original_col}"):
            norm_smiles = normalize_smiles_with_openbabel(smiles)
            normalized_list.append(norm_smiles)
            if norm_smiles:
                valid_count += 1
        
        normalized_df[normalized_col] = normalized_list
        stats[original_col] = (valid_count, len(normalized_list))
        print(f"✅ Успешно: {valid_count}/{len(normalized_list)} молекул")
    
    # Сохраняем нормализованные SMILES
    print(f"\n💾 Сохранение нормализованных SMILES в: {output_normalized_file}")
    normalized_df.to_csv(output_normalized_file, index=False)
    print(f"✅ Файл сохранен: {normalized_df.shape[0]} строк")
    
    print("\n🧠 === Этап 2: Генерация Morgan fingerprints ===")
    
    # Подготовка результирующего DataFrame
    result_df = df.copy()
    nBits = 2048
    radius = 2
    print(f"⚙️  Параметры fingerprint: radius={radius}, nBits={nBits}")
    
    # Генерация fingerprints для каждого типа молекул
    fingerprint_configs = [
        ('lig_1_normalized', 'L1', 'Morgan_1_'),
        ('lig_2_normalized', 'L2', 'Morgan_2_'),
        ('lig_3_normalized', 'L3', 'Morgan_3_'),
        ('solvent_normalized', 'Solvent_SMILES', 'Morgan_solv_abs_')
    ]
    
    total_bits = 0
    for norm_col, orig_col, prefix in fingerprint_configs:
        if norm_col not in normalized_df.columns:
            print(f"⏭️  Пропуск '{orig_col}': нет нормализованных данных")
            continue
            
        print(f"\n🧪 Генерация fingerprints для '{orig_col}'...")
        fps = []
        
        # Генерация векторов
        for smiles in tqdm(normalized_df[norm_col], desc=f"Fingerprint {prefix.strip('_')}"):
            fps.append(smiles_to_morgan(smiles, radius, nBits))
        
        # Добавление колонок в DataFrame
        print(f"עמוד Добавление {nBits} битовых колонок...")
        for bit_idx in range(nBits):
            col_name = f"{prefix}{bit_idx}"
            result_df[col_name] = [int(fp[bit_idx]) for fp in fps]
        total_bits += nBits
    
    # === ФИНАЛЬНАЯ СТАТИСТИКА ===
    print("\n🎉 === Обработка завершена! ===")
    print(f"📈 Итоговый DataFrame: {result_df.shape[0]} строк, {result_df.shape[1]} колонок")
    print(f"🧬 Создано fingerprint-колонок: {total_bits}")
    
    # Детальная статистика по колонкам
    print("\n📊 Детальная статистика нормализации:")
    for col, (valid, total) in stats.items():
        print(f"  - {col}: {valid}/{total} ({valid/total:.1%})")
    
    # Проверка качества нормализации
    invalid_count = sum(1 for col in normalized_df.columns if col.endswith('_normalized') 
                        for val in normalized_df[col] if val is None)
    if invalid_count:
        print(f"\n⚠️  Внимание: обнаружено {invalid_count} некорректных SMILES! "
              f"Проверьте файл '{output_normalized_file}'")
    
    return result_df, normalized_df

# Пример использования
if __name__ == "__main__":
    # Загрузка данных (пример)
    print("🧪 Демонстрация работы супер-функции нормализации:")
    test_smiles = [
        'C1=CC=[C]C(=C1)c1c2ccccc2ccn1',  # Радикал в цикле
        '[O-]N=O',                         # Простая соль
        '[CH3][C-2]=O',                    # Аномальный формальдегид
        '[2H]C([3H])([4H])C',              # Изотопы
        'CCO.[Pb]',                        # Тяжелый металл
        'N[C@@H](C)C(=O)O'                 # Валидная аминокислота
    ]
    
    print("\n🔍 Тестирование нормализации:")
    for smi in test_smiles:
        result = normalize_smiles_with_openbabel(smi)
        status = "✅ Валидный" if result else "❌ Невалидный"
        print(f"  {smi:25s} → {status} → {result or 'None'}")
    
    print("\n💡 Инструкция по использованию:")
    print("1. Загрузите данные: df = pd.read_csv('ваш_файл.csv')")
    print("2. Запустите обработку: result_df, norm_df = process_smiles_and_generate_fps(df)")
    print("3. Сохраните результаты: result_df.to_csv('результаты.csv', index=False)")
    
    print("\n⚡ Рекомендации для больших датасетов:")
    print("   - Используйте nBits=1024 для ускорения")
    print("   - Запускайте обработку на GPU-сервере")
    print("   - Разбивайте данные на чанки при работе с >1M строк")
```
